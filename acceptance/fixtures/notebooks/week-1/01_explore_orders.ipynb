# Week 1: Explore the orders

We look at a year of shop orders, to find out why some customers stop buying. This notebook loads the raw data, checks it, and saves a clean copy for next week.


## Load the raw data

Orders arrive as one file a month. We read them into a single table.


```python
import pandas as pd

orders = pd.read_csv('orders_raw.csv')
orders.head()
```

## Check the quality

Missing customer ids can't be used, so we drop those rows and count how many we lost.


```python
before = len(orders)
orders = orders.dropna(subset=['customer_id'])
print(before - len(orders), 'rows dropped')
```

## Save a clean copy

Week 2 starts from this file.


```python
orders.to_parquet('orders_clean.parquet')
```
