# Week 3: Evaluate it

Does the score predict who actually left? We use the service's code directly, and compare it with what happened.


## Score last quarter

We score every customer as they were three months ago.


```python
import sys

sys.path.append('../week-2')

import pandas as pd
from churn.scoring import risk_band, score_customer

clean = pd.read_parquet('../week-1/orders_clean.parquet')
scored = clean.assign(score=clean.days_since_order.map(score_customer))
```

## Compare with what happened

A good score puts most leavers in the high band.


```python
scored['band'] = scored.score.map(risk_band)
scored.groupby('band').left.mean()
```

```python
scored.to_csv('evaluation.csv')
```
